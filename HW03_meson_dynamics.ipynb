## Simulation of real-time "meson" confinement dynamics
Adapted from Kormos, M., Collura, M., Takács, G. et al. Real-time confinement following a quantum quench to a non-integrable model. Nature Phys 13, 246–249 (2017). https://doi.org/10.1038/nphys3934

At the subatomic level, the protons and neutrons are composed of quarks, whose interactions are described by the *strong force*. A peculiar property of the strong force is **confinement**: the further apart the quarks are, the stronger the force between them. In a loose sense, one can view the strong force as effectively introducing a linear potential between a quark and an antiquark. A bound state of a quark–antiquark pair is called a meson. Due to the effective linear potential between the quarks, pulling a meson apart requires injecting a large amount of energy — eventually enough to create a new quark–antiquark pair from the vacuum, so that two mesons have less energy than two separated quarks.

Interestingly, an analogous confinement phenomenon also shows up in a 1D mixed field quantum Ising chain. 
$$
H
=
-(J\sum_{j=0}^{n-1}X_jX_{j+1}
+h_z\sum_{j=0}^{n-1}Z_j
+h_x\sum_{j=0}^{n-1}X_j)~,
$$
when $h_x \ll J, h_z$.
This can be seen more easily by considering the limit $h_z = 0$, so everything can be described classically in the X-basis $|\pm\rangle$. Recall that when $h_x = 0$, we have spontaneous symmetry breaking, and the degenerate ground states are $|{+}{+}\ldots{+}\rangle$ and $|{-}{-}\ldots{-}\rangle$. Let us view these two states as "vacua". The elementary excitations of these vacua are not described by a single spin flip. They are, however, described by "domain walls" — each $+-$ or $-+$ configuration has additional energy $2J$. And one can see that $|{+}{+}\cdots{+}{-}{+}\cdots{+}\rangle$ has the same energy as $|{+}{+}\cdots{+}{-}\cdots{-}{+}\cdots{+}\rangle$. This tells us the domain walls $+-$ and $-+$ are the elementary excitations of the ground state. In a more fancy way, we say these are the "elementary particles" (if you are a particle physicist) or "quasi-particles" (if you are a condensed matter physicist) of the vacua.

Now if we introduce nonzero $h_x$, the vacuum $|{+}{+}\cdots{+}\rangle$ has energy $E_0 = -(J + h_x)N$; $|{+}{+}\cdots{+}{-}{+}\cdots{+}\rangle$ has an energy of $(4J + 2h_x) + E_0$, and $|{+}{+}\cdots{+}\underbrace{{-}\cdots{-}}_{\ell}{+}\cdots{+}\rangle$ has an energy of $(4J + 2h_x\ell) + E_0$, where $\ell$ is the distance between the two domain walls. You can see that $h_x$ introduces a linear potential between the domain walls. In analogy to the quarks and mesons, we call a pair of domain walls $+{-}{-}+$ a "meson", and the domain walls are confined due to the linear potential between them.

Of course, without introducing $h_z$, everything is just classical and can be solved exactly. The interesting part is that this confinement picture persists even when $h_z \neq 0$, as long as the vacua retain the property that the quasi-particles are described by (dressed) domain walls.

### Domain wall dynamics — spacetime visualization of confinement

Prepare the initial state
$$
|\psi_0\rangle = |{+}\cdots{+}{-}{-}{+}\cdots{+}\rangle
$$
(two domain walls separated by $\ell_0 = 2$ sites) on a chain of at least $n = 18$ qubits. Evolve it under the mixed-field Ising model with periodic boundary conditions.

Define the domain wall density at bond $j$ as
$$
d_j(t) = \frac{1 - \langle X_j X_{j+1}\rangle_t}{2},
$$
which equals 1 at an anti-aligned bond and 0 at an aligned bond. (The `expect_XX` function in the previous homework set might be helpful.)

1. (10pts) Consider the states $|\ell\rangle = |{+}{+}\cdots{+}\underbrace{{-}\cdots{-}}_{\ell}{+}\cdots{+}\rangle$. Calculate $E(\ell) = \langle \ell | H | \ell \rangle$ for several small $\ell$ with $h_z = 0.3$, $h_x = 0.3$. Use $E(\ell) = A + \sigma\ell$ to get the effective string tension $\sigma$.

2. (20pts) Find the ground state energy $E_0$ and the first excitation state energy $E_1$, and compute the energy of the initial state $E_\mathrm{ini} = \langle \psi_0 | H | \psi_0 \rangle$ for $h_z = 0.3$, $h_x = 0.3$. Since the first excitation state can be viewed as the two domain walls are exicted from the vaccum with zero kinetic energy, estimate the effective mass (in the unit of energy) of the domain wall $E_m$ via $2E_m = E_1 - E_0$. Find the kinetic energy in the initial state as 
$$
K := E_\mathrm{ini} - E_1 ~.
$$


3. (20pts) For $h_z = 0.3$, $h_x = 0$: plot a spacetime heatmap of $d_j(t)$ (bond $j$ on the $x$-axis, time $t$ on the $y$-axis). (Use your judgement to pick the resolution in $t$ and the total time $t = T$ for simulation.)
Describe the propagation pattern. What is the initial velocity of the domain walls?

4. (20pts) Repeat for $h_z = 0.3$, $h_x = 0.3$. Describe the qualitative difference. Do the domain walls escape?


5. (10pts) Consider two bodies connected by a string with constant tension $\sigma$ (a constant attractive force of magnitude $\sigma$). Derive the oscillation period $T$ as a function of $\sigma$, $K$, and their effective mass $m$. Using $\sigma$ and $K$ from problems 1–2, and estimating $m^*$ from the domain wall group velocity $v$ measured in problem 3 (hint: $K = 2* \frac{1}{2}m^*v^2$ so $m^* \approx K/v^2$), calculate $T$ and compare with the period observed in problem 4.
