# Quantum Mechanics & Linear Algebra Review

These notes establish the mathematical language used throughout the course. The goal is not to teach quantum mechanics from scratch, but to fix notation, connect the Dirac formalism to concrete matrix algebra, and introduce the computational vocabulary (bitstrings, Pauli operators, endianness) that becomes essential starting in Week 2.

## 1. Hilbert Space

A quantum state lives in a **Hilbert space** $\mathcal{H}$ — a complete complex vector space equipped with an inner product.

For a **single qubit**, $\mathcal{H} = \mathbb{C}^2$: states are two-component complex vectors.
For **$n$ qubits**, $\mathcal{H} = \mathbb{C}^{2^n}$: the dimension grows exponentially with system size.

The **inner product** $\langle \cdot, \cdot \rangle : \mathcal{H} \times \mathcal{H} \to \mathbb{C}$ satisfies:

| Property | Condition |
|---|---|
| Conjugate symmetry | $\langle \phi, \psi \rangle = \langle \psi, \phi \rangle^*$ |
| Linearity in 2nd argument | $\langle \phi,\, \alpha\psi + \beta\chi \rangle = \alpha\langle\phi,\psi\rangle + \beta\langle\phi,\chi\rangle$ |
| Positive-definiteness | $\langle \psi, \psi \rangle \geq 0$, with equality iff $\psi = 0$ |

A set of vectors $\{|e_i\rangle\}$ forms an **orthonormal basis** if $\langle e_i | e_j \rangle = \delta_{ij}$ and every vector in $\mathcal{H}$ can be written as a linear combination of them.

> **The exponential wall.** An $n$-qubit state vector has $2^n$ complex amplitudes. At $n = 30$, that is $\sim 10^9$ entries, requiring $\sim 16\,\text{GB}$ of memory (each `complex128` takes 16 bytes). This exponential scaling is the central difficulty of classical simulation — and the reason the rest of this course exists.

## 2. Bra-Ket Notation

Dirac's bra-ket notation separates the abstract quantum state from its representation in any particular basis.

- A **ket** $|\psi\rangle$ is a vector in $\mathcal{H}$.
- A **bra** $\langle\psi|$ is its dual vector (the conjugate transpose of $|\psi\rangle$).

In column/row vector form:
$$|\psi\rangle \;\leftrightarrow\; \begin{pmatrix}c_0 \\ c_1 \\ \vdots \\ c_{N-1}\end{pmatrix}, \qquad \langle\psi| \;\leftrightarrow\; \begin{pmatrix}c_0^* & c_1^* & \cdots & c_{N-1}^*\end{pmatrix}$$

where $c_i = \langle e_i | \psi \rangle$ are the components in the chosen basis.

**Inner product** (a scalar):
$$\langle\phi|\psi\rangle = \sum_i d_i^*\, c_i \;\in\; \mathbb{C}$$

**Outer product** (an operator):
$$|\psi\rangle\langle\phi| \quad \text{maps } |\chi\rangle \;\mapsto\; |\psi\rangle\,\langle\phi|\chi\rangle$$

**Normalization:** Physical states satisfy $\langle\psi|\psi\rangle = 1$.

**Completeness relation:** Any orthonormal basis $\{|e_i\rangle\}$ satisfies
$$\sum_i |e_i\rangle\langle e_i| = \mathbf{I}$$
Inserting this identity between operators or states is one of the most frequently used techniques in QM calculations.

## 3. Operators

An **operator** $\hat{O}$ on $\mathcal{H}$ is a linear map $\mathcal{H} \to \mathcal{H}$. In a chosen basis it becomes a matrix.

### Hermitian Operators

$\hat{O}$ is **Hermitian** if $\hat{O}^\dagger = \hat{O}$, where $^\dagger$ denotes the conjugate transpose. Key consequences:

- All eigenvalues are **real** — essential for physical observables.
- Eigenvectors for distinct eigenvalues are **orthogonal**.
- The eigenvectors form a **complete set**: any state can be expanded in them.

All physical observables (energy, spin components, particle number, ...) are Hermitian operators.

### Unitary Operators

$\hat{U}$ is **unitary** if $\hat{U}^\dagger \hat{U} = \hat{U}\hat{U}^\dagger = \mathbf{I}$. Key consequences:

- Preserves norms: $\langle U\psi | U\psi \rangle = \langle\psi|\psi\rangle$.
- All eigenvalues lie on the unit circle in $\mathbb{C}$: $|\lambda| = 1$.

Quantum time evolution and quantum gates are unitary operations.

## 4. Tensor Products and Composite Systems

When a quantum system is composed of $n$ subsystems (e.g., $n$ qubits), its Hilbert space is the **tensor product** of the individual spaces.

### The Tensor Product Space

For two qubits, each with $\mathcal{H} = \mathbb{C}^2$:
$$\mathcal{H}_{12} = \mathcal{H}_1 \otimes \mathcal{H}_2 \;\cong\; \mathbb{C}^4$$

For $n$ qubits:
$$\mathcal{H} = (\mathbb{C}^2)^{\otimes n} \;\cong\; \mathbb{C}^{2^n}$$

### Product States

If qubit 1 is in state $|\psi\rangle$ and qubit 2 is in state $|\phi\rangle$, the joint state is:
$$|\psi\rangle \otimes |\phi\rangle \;\equiv\; |\psi\phi\rangle$$

Concretely, if $|\psi\rangle = \begin{pmatrix}a\\b\end{pmatrix}$ and $|\phi\rangle = \begin{pmatrix}c\\d\end{pmatrix}$, then:
$$|\psi\rangle \otimes |\phi\rangle = \begin{pmatrix}ac\\ad\\bc\\bd\end{pmatrix}$$

The Kronecker product gives the rule: row $i$ of $|\psi\rangle$ multiplies all of $|\phi\rangle$.

### Entangled States

A state that **cannot** be written as $|\psi_1\rangle \otimes \cdots \otimes |\psi_n\rangle$ is **entangled**. The classic example is the Bell state:
$$|\Phi^+\rangle = \frac{1}{\sqrt{2}}\bigl(|00\rangle + |11\rangle\bigr)$$
No single-qubit states $|\psi\rangle$, $|\phi\rangle$ satisfy $|\psi\rangle \otimes |\phi\rangle = |\Phi^+\rangle$.

Entanglement is the key quantity distinguishing quantum from classical systems, and its structure determines which numerical methods are efficient. This theme appears throughout the course.

### Operators on Composite Systems

If $A$ acts on qubit 1 and $B$ acts on qubit 2, the joint operator on $\mathcal{H}_{12}$ is $A \otimes B$:
$$(A \otimes B)\,(|\psi\rangle \otimes |\phi\rangle) = (A|\psi\rangle) \otimes (B|\phi\rangle)$$

To apply $A$ to qubit 1 only, use $A \otimes I$. To apply $B$ to qubit 2 only, use $I \otimes B$.

## 5. Vector and Matrix Representations

Choosing an orthonormal basis $\{|e_i\rangle\}$ makes everything concrete and computable.

### State as a Column Vector

$$|\psi\rangle = \sum_{i=0}^{N-1} c_i\,|e_i\rangle \quad\longleftrightarrow\quad \mathbf{c} = \begin{pmatrix}c_0 \\ c_1 \\ \vdots \\ c_{N-1}\end{pmatrix}, \qquad c_i = \langle e_i|\psi\rangle$$

The bra $\langle\psi|$ is the conjugate-transpose row vector $\mathbf{c}^\dagger = (c_0^*,\, c_1^*,\, \ldots,\, c_{N-1}^*)$.

### Operator as a Matrix

An operator $\hat{O}$ is represented by the $N \times N$ matrix with entries:
$$O_{ij} = \langle e_i|\,\hat{O}\,|e_j\rangle$$

Under this representation:

| Abstract | Matrix/vector form |
|---|---|
| Operator composition $\hat{A}\hat{B}$ | Matrix product $AB$ |
| Hermitian conjugate $\hat{O}^\dagger$ | Conjugate transpose $(O^*)^T$ |
| Action on state $\hat{O}|\psi\rangle$ | Matrix-vector product $O\mathbf{c}$ |
| Inner product $\langle\phi|\psi\rangle$ | Dot product $\mathbf{d}^\dagger \mathbf{c}$ |
| Expectation value $\langle\psi|\hat{O}|\psi\rangle$ | $\mathbf{c}^\dagger O\, \mathbf{c}$ |

## 6. Computational Basis and Bitstrings

For $n$ qubits, the standard basis is the **computational basis**: the $2^n$ product states that are simultaneous eigenstates of $Z_0, Z_1, \ldots, Z_{n-1}$.

### Basis Labeling

Each computational basis state is labeled by a **bitstring** $s = s_0 s_1 \cdots s_{n-1}$ with $s_k \in \{0,1\}$:
$$|s\rangle = |s_0\rangle \otimes |s_1\rangle \otimes \cdots \otimes |s_{n-1}\rangle$$

The bitstring maps to an integer index (qubit 0 is the most significant bit):
$$\text{index}(s) = \sum_{k=0}^{n-1} s_k \cdot 2^{n-1-k}$$

For $n = 3$:

| Bitstring $s_0 s_1 s_2$ | Index |
|:---:|:---:|
| 000 | 0 |
| 001 | 1 |
| 010 | 2 |
| 011 | 3 |
| 100 | 4 |
| 101 | 5 |
| 110 | 6 |
| 111 | 7 |

### State Vector in Code

A general $n$-qubit state $|\psi\rangle = \sum_s c_s |s\rangle$ is stored as a complex array of length $2^n$, where `psi[i]` holds the amplitude $c_s$ of the basis state with index $i$.

## 7. Pauli Operators

The four **Pauli matrices** are the fundamental single-qubit operators:

$$I = \begin{pmatrix}1&0\\0&1\end{pmatrix}, \quad X = \begin{pmatrix}0&1\\1&0\end{pmatrix}, \quad Y = \begin{pmatrix}0&-i\\i&0\end{pmatrix}, \quad Z = \begin{pmatrix}1&0\\0&-1\end{pmatrix}$$

### Physical Interpretation

- $Z$ has eigenstates $|0\rangle$ (eigenvalue $+1$) and $|1\rangle$ (eigenvalue $-1$) — it defines the computational basis.
- $X$ is the **bit-flip**: $X|0\rangle = |1\rangle$, $X|1\rangle = |0\rangle$.
- $Y$ combines bit- and phase-flip: $Y|0\rangle = i|1\rangle$, $Y|1\rangle = -i|0\rangle$.

### Key Properties

| Property | Statement |
|---|---|
| Hermitian | $X^\dagger = X,\quad Y^\dagger = Y,\quad Z^\dagger = Z$ |
| Unitary / involutory | $X^2 = Y^2 = Z^2 = I$ |
| Traceless | $\operatorname{tr}(X) = \operatorname{tr}(Y) = \operatorname{tr}(Z) = 0$ |
| Anticommutation | $XY + YX = 0$ (and cyclic permutations) |
| Multiplication rule | $XY = iZ,\quad YZ = iX,\quad ZX = iY$ |

Together with $I$, the Paulis $\{I, X, Y, Z\}$ span the space of all $2\times 2$ complex matrices.

## 8. Endianness

When we label a multi-qubit state $|s_0 s_1 \cdots s_{n-1}\rangle$, we must decide which qubit is "qubit 0" and how bitstrings map to array indices. This is the quantum analogue of byte endianness in classical computing, and it is a persistent source of bugs when interfacing different tools.

### The Two Common Conventions

**Physics convention — big-endian (used in this course):**
Qubit 0 is the **leftmost** (most significant) bit.
$$\text{index}(s_0 s_1 \cdots s_{n-1}) = s_0\cdot 2^{n-1} + s_1\cdot 2^{n-2} + \cdots + s_{n-1}\cdot 2^0$$

Example ($n=3$): $|100\rangle$ → index 4; $|001\rangle$ → index 1.

**Qiskit convention — little-endian:**
Qubit 0 is the **rightmost** (least significant) bit.
$$\text{index}(s_0 s_1 \cdots s_{n-1}) = s_0\cdot 2^0 + s_1\cdot 2^1 + \cdots + s_{n-1}\cdot 2^{n-1}$$

Example ($n=3$): $|100\rangle$ → index 1; $|001\rangle$ → index 4.

### Why It Matters

Endianness affects which matrix row/column corresponds to which basis state, the order of tensor products when building $n$-qubit operators, and comparisons with results from quantum hardware or other software.

**Always state your convention explicitly.** In this course we use the physics (big-endian) convention throughout.

## 9. Pauli Strings

A **Pauli string** on $n$ qubits is a tensor product of $n$ single-qubit Pauli operators, each drawn from $\{I, X, Y, Z\}$. For example, on 4 qubits:
$$X \otimes Z \otimes I \otimes Y \;\equiv\; XZIY$$

### Basis for Operator Space

The $4^n$ Pauli strings on $n$ qubits form a **complete orthonormal basis** for the space of $2^n \times 2^n$ complex matrices (under the Hilbert-Schmidt inner product $\langle A, B\rangle = \operatorname{tr}(A^\dagger B)$). This means any operator can be expanded as:
$$\hat{O} = \sum_{\alpha} o_\alpha\, P_\alpha$$

For a Hermitian operator (e.g., a Hamiltonian), all coefficients $o_\alpha$ are real.

### Example: Transverse-Field Ising Model

The benchmark Hamiltonian used throughout this course:
$$H_\text{TFIM} = -J\sum_{i=0}^{n-2} Z_i Z_{i+1} \;-\; h\sum_{i=0}^{n-1} X_i$$

Here $Z_i \equiv I^{\otimes i} \otimes Z \otimes I^{\otimes(n-1-i)}$ means the Pauli $Z$ acting on qubit $i$ and the identity on all other qubits. Each term in the sum is a Pauli string.

### Relevance for the Course

Pauli strings appear in all three simulation paradigms:
- **ED (Unit 2):** The Hamiltonian matrix is built by summing Pauli string matrices.
- **MPS (Unit 5):** Hamiltonians are encoded as matrix product operators (MPOs), one term per Pauli string.
- **Heisenberg methods (Unit 6):** Time evolution is tracked by propagating Pauli strings forward in operator space.

## 10. The Time-Independent Schrödinger Equation

For a system with Hamiltonian $H$ (a Hermitian operator), the **time-independent Schrödinger equation** is the eigenvalue problem:
$$H|\psi\rangle = E|\psi\rangle$$

The allowed energies $E$ are the eigenvalues of $H$; the corresponding states $|\psi\rangle$ are the energy eigenstates.

Since $H$ is Hermitian:
- All energies $E_k$ are **real**.
- Eigenstates for distinct energies are **orthogonal**: $\langle E_j | E_k\rangle = \delta_{jk}$.
- The eigenstates span $\mathcal{H}$: any state can be written $|\psi\rangle = \sum_k c_k |E_k\rangle$.

The smallest eigenvalue $E_0$ is the **ground state energy** and $|E_0\rangle$ is the **ground state**. Finding $E_0$ and $|E_0\rangle$ efficiently — for systems too large to diagonalize exactly — is one of the central problems of computational quantum physics.

## 11. Solving the TISE by Diagonalization

In the computational basis, $H$ becomes a $2^n \times 2^n$ Hermitian matrix. **Diagonalization** finds the unitary $V$ and diagonal matrix $\Lambda$ such that:
$$H = V\,\Lambda\,V^\dagger, \qquad \Lambda = \operatorname{diag}(E_0,\, E_1,\, \ldots,\, E_{2^n-1})$$

The $k$-th column of $V$ is the eigenvector $|E_k\rangle$; the $k$-th diagonal entry of $\Lambda$ is the energy $E_k$.

### Computational Cost

Full diagonalization of a $D \times D$ matrix costs $\mathcal{O}(D^3)$. For $n$ qubits, $D = 2^n$:

| $n$ | $D = 2^n$ | $\mathcal{O}(D^3)$ |
|:---:|:---:|:---:|
| 10 | 1,024 | $\sim 10^9$ — fast |
| 16 | 65,536 | $\sim 2.8 \times 10^{14}$ — slow |
| 20 | $\sim 10^6$ | intractable |

Full diagonalization is feasible on a laptop only for $n \lesssim 14$–16.

### In Practice

Two facts allow us to go further:
1. $H$ is typically **sparse** — most entries are zero (each Pauli string term touches only a few matrix elements).
2. We usually need only the **ground state** (or a few lowest eigenstates), not all $2^n$.

Iterative methods (Lanczos, covered in Unit 4) exploit sparsity to reach $n \sim 24$–30.

## 12. Expectation Values and Measurement

Given a normalized state $|\psi\rangle$ and an observable $\hat{O}$ (Hermitian), the **expectation value** is:
$$\langle\hat{O}\rangle_\psi = \langle\psi|\,\hat{O}\,|\psi\rangle$$

This is the average measurement outcome over many identically prepared copies of $|\psi\rangle$.

### Matrix-Vector Form

$$\langle\hat{O}\rangle = \mathbf{v}^\dagger\, O\, \mathbf{v}$$

where $\mathbf{v}$ is the column vector for $|\psi\rangle$ and $O$ is the matrix for $\hat{O}$.

### Expansion in the Eigenbasis

If $\hat{O}$ has eigenstates $|\lambda_k\rangle$ with eigenvalues $\lambda_k$, and we expand:
$$|\psi\rangle = \sum_k c_k\,|\lambda_k\rangle$$
then:
$$\langle\hat{O}\rangle = \sum_k \lambda_k\,|c_k|^2$$

The values $|c_k|^2$ are probabilities (they sum to 1 by normalization), and $\lambda_k$ are the only possible individual measurement outcomes.

### Common Observables in This Course

| Observable | Expression | Physical meaning |
|---|---|---|
| Local magnetization | $\langle Z_i \rangle$ | Average spin at site $i$ |
| Two-point correlator | $\langle Z_i Z_j \rangle$ | Spin-spin correlation |
| Energy | $\langle H \rangle$ | Expectation of the Hamiltonian |
| Variance | $\langle H^2\rangle - \langle H\rangle^2$ | Energy spread; zero for an eigenstate |

Expectation values are the primary output of every simulation in this course — the bridge between the wavefunction and measurable physics.
