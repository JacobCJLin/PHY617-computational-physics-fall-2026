### The Quantum Ising Model

The (classical) Ising model is one of the simplest and most influential models in statistical and condensed-matter physics. It was introduced in the early 1920s by Wilhelm Lenz as a simplified model of ferromagnetism and was first analyzed by his student [Ernst Ising](https://en.wikipedia.org/wiki/Ernst_Ising). In the model, each microscopic magnetic moment is represented by $\sigma = \pm 1$. Ising solved the one-dimensional classical model and found that it does not exhibit a finite-temperature ferromagnetic phase transition. The importance of the model became clear later, particularly after it was shown that the two-dimensional version does support a phase transition and after [Onsager](https://en.wikipedia.org/wiki/Lars_Onsager) obtained its exact solution. Since then, the Ising model has become a central paradigm for understanding collective behavior, spontaneous symmetry breaking, critical phenomena, and universality.

The quantum Ising model is obtained by replacing the classical spins with quantum-mechanical spin-$1/2$ operators, with an additional transverse field.

We consider the one-dimensional transverse-field Ising model,
$$
H
=
-J\sum_{j=0}^{n-1}X_jX_{j+1}
-h\sum_{j=0}^{n-1}Z_j~,
$$
where we identify $j+n \equiv j$ (periodic boundary conditions). The interaction proportional to $J$ favors alignment of neighboring spins along the $x$ direction (ferromagnetic interaction), while the transverse field $h$ favors polarization along the $z$ direction. These two terms do not commute and therefore represent competing tendencies that cannot be minimized simultaneously.
The model has a $\mathbb{Z}_2$ symmetry generated by $P_{\mathbb{Z}_2} = \prod_j Z_j$, as we saw in the in-class notebook.



### 1. Spontaneous symmetry breaking

When $J\gg h$, the interaction dominates and the ground state develops ferromagnetic order along the $x$ direction. In the **thermodynamic limit** (system size $\rightarrow \infty$), the system spontaneously chooses one of two symmetry-related ordered states from an infinitesimal small symmetry-breaking perturbation. In fact, this can be seen at the point $h=0$ exactly --- there are two degenerate ground states $|{+}{+}\cdots{+}\rangle$ and $|{-}{-}\cdots{-}\rangle$, where $|\pm\rangle = (|0\rangle \pm |1\rangle)/\sqrt{2}$ are the $\pm 1$ eigenstates of $X$. These two states are related by $P_{\mathbb{Z}_2} = \prod_j Z_j$ (which maps $|\pm\rangle \mapsto |\mp\rangle$), and choosing one of them spontaneously *breaks* the $\mathbb{Z}_2$ symmetry. Of course, since these two ground states are degenerate, one can also form eigenstates which respect the $\mathbb{Z}_2$ symmetry,
$$
|\Psi_\pm\rangle \propto |{+}{+}\cdots{+}\rangle \pm |{-}{-}\cdots{-}\rangle.
$$

This is in fact a salient feature — perhaps the definition — of spontaneous symmetry breaking: the ground state is degenerate, and while one can form symmetry-respecting superpositions $|\Psi_\pm\rangle$, these are exponentially fragile in the thermodynamic limit. The degeneracy coming from different symmetry sectors: The ground state from the $Z_2 = +1$ sector and the ground state from the $Z_2 = -1$ sector become exponentially close in energy, with a splitting that vanishes exponentially with system size. This allows one to form symmetry-breaking states from the nearly degenerate ground-state subspace.
An infinitesimally small symmetry-breaking perturbation therefore can select one of the symmetry-broken states $|{+}{+}\cdots{+}\rangle$ or $|{-}{-}\cdots{-}\rangle$.

In contrast, when $h\gg J$, the transverse field dominates and the spins align approximately along the $z$ direction, producing a quantum paramagnetic phase, and the ground state is unique and respecting the $Z_2$ symmetry. 
Again, we can examine the extreme case where $J=0$. In this case, there is only one ground state $|00\cdots0\rangle$, which respects the $\mathbb{Z}_2$ symmetry.

The competition between these two limiting cases gives rise to a *quantum phase transition*. For the one-dimensional nearest-neighbor model, the critical point occurs at
$$
h_c=J.
$$
Unlike an ordinary thermal phase transition, which is driven by thermal fluctuations, this transition occurs at zero temperature and is driven entirely by quantum fluctuations.

### 2. Order parameter

To diagnose spontaneous symmetry breaking, the famous Soviet physicist [Lev Landau](https://en.wikipedia.org/wiki/Lev_Landau) introduced the concept of an *order parameter*. In the transverse-field Ising model, we know from physics that the magnetization should be nonzero in the *ferromagnetic phase*, while being zero in the *paramagnetic phase*.

The natural order parameter is therefore the average magnetization along the $x$ direction,
$$
M_x = \frac{1}{n}\sum_{j=0}^{n-1} \langle X_j \rangle.
$$
In the thermodynamic limit, $M_x \neq 0$ in the ordered phase ($J \gg h$) and $M_x = 0$ in the disordered phase ($h \gg J$).

More generally, the order parameter is chosen to be an operator that transforms *nontrivially* under the symmetry — in the jargon of condensed matter physics, it is **charged** under the symmetry. In our case, $P_{\mathbb{Z}_2}\, X_j\, P_{\mathbb{Z}_2} = -X_j$, so $M_x \to -M_x$ under $\mathbb{Z}_2$. This immediately implies $\langle M_x\rangle = 0$ in any $\mathbb{Z}_2$-symmetric state (as in a finite system), while $\langle M_x\rangle \neq 0$ only when the symmetry is broken — which is exactly what makes it a useful diagnostic.

However, if you think about it more carefully, you would spot a problem --- numerically, if we obtain the ground states which respect the symmetry at any finite sizes, we see $P_{\mathbb{Z}_2}|E_0\rangle = \pm |E_0\rangle$, then we have $\langle E_0 | \sum_i X_i | E_0 \rangle = \langle E_0 |P_{\mathbb{Z}_2} \sum_i X_i P_{\mathbb{Z}_2}| E_0 \rangle = -\langle E_0 | \sum_i X_i | E_0 \rangle = 0$.
The remedy is to use the disconnected spin correlation function instead,
$$
C(m,n) = \langle E_0|\, X_m X_n\,|E_0 \rangle~,
$$
in the limit $|m-n|\rightarrow \infty$.

### 3. Why the interest?

The one-dimensional transverse-field Ising model is especially important because it is both physically rich and exactly solvable. Through the [Jordan-Wigner transformation](https://en.wikipedia.org/wiki/Jordan–Wigner_transformation), the spin Hamiltonian can be mapped onto a model of noninteracting fermions. This mapping allows the energy spectrum, excitation gap, correlation functions, and thermodynamic properties to be obtained analytically. It therefore becomes a very popular model for benchmarking various numerical methods due to its exact solvability.

Despite its apparent simplicity, the model captures many of the central concepts of quantum many-body physics, including spontaneous symmetry breaking, quasiparticle excitations, diverging correlation lengths, finite-size scaling, and quantum criticality.

The quantum Ising model is not merely a theoretical construction. The quasi-one-dimensional magnetic material $\mathrm{CoNb}_2\mathrm{O}_6$ closely resembles an Ising chain in a transverse field; experiments on this material have observed quantum-critical behavior and signatures of an emergent $E_8$ excitation spectrum described by Zamolodchikov's Ising field theory. (In HW3, we will simulate the confinement dynamics of the “mesons” formed by confined domain-wall excitations using a quantum Ising chain lattice model.) Variants of the Ising Hamiltonian are also implemented in programmable quantum simulators: trapped ions naturally realize long-range Ising interactions, while arrays of Rydberg atoms, superconducting qubits, and other platforms can reproduce related spin Hamiltonians, making it possible to study equilibrium phase transitions as well as nonequilibrium dynamics such as quantum quenches.

By varying $h/J$ and comparing several system sizes, you will identify finite-size signatures of the ferromagnetic and paramagnetic phases and of the quantum critical point separating them. Because a finite system cannot exhibit true spontaneous symmetry breaking, the exact ground state preserves the $\mathbb{Z}_2$ symmetry and $\langle X_j\rangle = 0$ identically. Finite-size signatures of the transition are therefore best seen in the excitation gap (which closes at $h_c$ in the thermodynamic limit), the spin correlation function $\langle X_i X_j\rangle$, and the variance of the order parameter $M_x$.

### Problem 3: Entanglement entropy (25 pts)

As a teaser for the future topics, here we examine the entaglement entropy of the ground state of the Ising model.

Paritition the system into a contiguous block $A$ of $\ell$ sites and its complement $B$ of $n-\ell$ sites.
For a normalized ground state $|\psi_0\rangle$, the reduced density matrix and the
**von Neumann entanglement entropy** are defined as 
$$
\rho_A = \mathrm{Tr}_B\, |\psi_0\rangle\langle\psi_0|~,
\qquad
S(\ell) = -\mathrm{Tr}\,\rho_A \ln \rho_A~.
$$

To calculate the entanglement entropy, you actually don't need to build $\rho_A$ explicitly. Reshaping the $2^n$ amplitude vector into a
$2^\ell \times 2^{n-\ell}$ matrix $M$ and taking its **[singular value decomposition](https://en.wikipedia.org/wiki/Singular_value_decomposition)**
$M = U\,\mathrm{diag}(\lambda_k)\,V^\dagger$ gives the Schmidt coefficients directly, and
$$
S(\ell) = -\sum_k \lambda_k^2 \ln \lambda_k^2~.
$$



Entanglement scaling:

1. **Constant scaling.** The ground state of a *gapped* local 1D Hamiltonian has $S(\ell)$ that saturates
   to a constant once $\ell \gtrsim \xi$ — it stops growing with block size. This is why
   matrix-product states / DMRG work so well away from criticality.
2. **Logarithmic scaling.** At $h=h_c$ the theory is a conformal field theory and
   Calabrese and Cardy showed that for a block of $\ell$ sites in a *periodic* chain of $n$ sites
   $$
   S(\ell) = \frac{c}{3}\,\ln\!\left[\frac{n}{\pi}\sin\!\frac{\pi \ell}{n}\right] + c_1~,
   $$
   where $c$ is the **central charge** of the CFT and $c_1$ is a non-universal constant. The
   quantum Ising transition sits in the Ising universality class, for which $c = 1/2$.
